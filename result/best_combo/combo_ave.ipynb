```python
import pandas as pd

df = pd.read_excel('combo_results.xlsx')

method_averages = {}

method_averages['데이터_선택'] = df.groupby('데이터_선택')['F1_점수'].mean().sort_values(ascending=False)
method_averages['피처_방법'] = df.groupby('피처_방법')['F1_점수'].mean().sort_values(ascending=False)
method_averages['모델_방법'] = df.groupby('모델_방법')['F1_점수'].mean().sort_values(ascending=False)
method_averages['리샘플링'] = df.groupby('리샘플링')['F1_점수'].mean().sort_values(ascending=False)
method_averages['손실함수'] = df.groupby('손실함수')['F1_점수'].mean().sort_values(ascending=False)

for method_type, averages in method_averages.items():
    print(f"\n=== {method_type} 별 평균 F1 점수 ===")
    for method, score in averages.items():
        print(f"{method:20} : {score:.3f}")

# 전체 방법별 평균을 하나의 DataFrame으로 생성
summary_data = []
for method_type, averages in method_averages.items():
    for method, score in averages.items():
        summary_data.append({
            '방법_유형': method_type,
            '방법': method,
            '평균_F1점수': round(score, 3)
        })

summary_df = pd.DataFrame(summary_data)

# 피벗 테이블로 변환
pivot_table = summary_df.pivot(index='방법', columns='방법_유형', values='평균_F1점수')
print(f"\n=== 전체 방법별 평균 요약 ===")
pivot_table.fillna('-')
```
